**Task 3 – Flattening Concept**
- Flatten traffic dataset


- Flatten sales dataset


- Compare behavior of:
- flatten()
- ravel()


```python
fal=traffic.flatten()
fla_sales=sales.flatten()
ravel=traffic.ravel()
ravel_sales=sales.ravel()
print("Flatten traffic dataset\n:",fal)
print("flatten sales dataset\n:",fla_sales)
print("ravel traffic dataset\n:",ravel)
print("ravel sales dataset\n:",ravel_sales)
```

**flatten():**

- Returns a new copy of all 480 elements in 1D.

- Shape (480,).

- Safe to modify without touching the original sales dataset.

**ravel():**

- Returns a view of the same 480 elements in 1D.

- Shape (480,).

- Modifications propagate back to the original sales dataset.
